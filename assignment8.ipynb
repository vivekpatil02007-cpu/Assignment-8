```python
import heapq

def astar(graph, heuristic, start, goal):
    # Priority queue stores (f(n), current_cost, node)
    pq = [(heuristic[start], 0, start)]

    # Store the parent of each node
    parent = {start: None}

    # Store the minimum cost from start to each node
    g_cost = {start: 0}

    while pq:
        f, cost, current = heapq.heappop(pq)

        # Skip outdated queue entries
        if cost != g_cost[current]:
            continue

        # Check whether the goal is reached
        if current == goal:
            path = []
            node = goal

            while node is not None:
                path.append(node)
                node = parent[node]

            path.reverse()
            print("Path:", path)
            print("Total Cost:", g_cost[goal])
            return path

        # Explore neighboring nodes
        for neighbor, weight in graph[current].items():
            new_cost = g_cost[current] + weight

            if neighbor not in g_cost or new_cost < g_cost[neighbor]:
                g_cost[neighbor] = new_cost
                parent[neighbor] = current

                f_cost = new_cost + heuristic[neighbor]
                heapq.heappush(
                    pq, (f_cost, new_cost, neighbor)
                )

    print("No path found!")
    return None


# Graph given in the experiment
graph = {
    'A': {'B': 2, 'C': 5},
    'B': {'A': 2, 'D': 2, 'E': 6},
    'C': {'A': 5, 'F': 3},
    'D': {'B': 2},
    'E': {'B': 6, 'F': 1},
    'F': {'C': 3, 'E': 1}
}

# Heuristic values given in the experiment
heuristic = {
    'A': 7,
    'B': 5,
    'C': 3,
    'D': 3,
    'E': 1,
    'F': 0
}

# Starting and goal nodes
start = 'A'
goal = 'F'

# Run the A* algorithm
astar(graph, heuristic, start, goal)
```
